```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from datetime import datetime

todays_date = datetime.now().strftime("%Y-%m-%d")

df_new = pd.read_excel('C:\\Files\\Excel\\Project\\Happiness_Development_Index.xlsx')
df_new.columns= df_new.columns.str.strip()
student_id = "Arnjon0191"

current_datetime = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

#Create Histogram of HDI Scores
plt.figure(figsize=(10, 6))
plt.hist(df_new['Human Development Index (HDI)'], bins=20, color='skyblue', edgecolor='black')
plt.xlabel('Human Development Index (HDI)')
plt.ylabel('Number of Countries')
plt.title(f'Histogram of HDI Scores - StudentID: {student_id} - {current_datetime}')
plt.grid(True)
plt.show()
```

```python
#Top Ten countries
top_10_countries = df_new.nlargest(10, 'Human Development Index (HDI)')

#Create Bar
plt.figure(figsize=(12, 8))
plt.barh(top_10_countries['Country'],
top_10_countries['Human Development Index (HDI)'], color='red')
plt.xlabel('HDI')
plt.ylabel('Country')
plt.title(f'Top 10 Countries by HDI  - StudentID: {student_id} - {current_datetime}')
plt.gca().invert_yaxis()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

#Define Variables
hdi = df_new["Human Development Index (HDI)"]
gni = df_new["Gross national income (GNI) per capita"]

#Create Scatter
plt.figure(figsize=(12,8))

plt.scatter(gni, hdi, color="purple")

plt.xlabel("Gross national income (GNI) per capita")
plt.ylabel("Human Development Index (HDI)")
plt.title(f"HDI Vs GNI per Capita - StudentID: {student_id} - {current_datetime}")

plt.grid(True)
plt.show()
```

```python
selected_columns = ['Human Development Index (HDI)', 'Life expectancy at birth', 'Expected years of schooling', 'Mean years of schooling', 'Gross national income (GNI) per capita']
correlation_matrix = df_new[selected_columns].corr()
plt.figure(figsize=(10, 8))

#Create Heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title(f'Heatmap of Correlations - StudentID: {student_id} - {current_datetime}')
plt.show()
```
